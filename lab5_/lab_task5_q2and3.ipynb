```python
#Question 2
#Minimax with Leaf Evaluation Counter
import math
count = 0
def minimax(curDepth, nodeIndex, maxTurn, scores, targetDepth):
    global count
    if curDepth == targetDepth:
        count += 1
        return scores[nodeIndex]
    if maxTurn:
        return max(
            minimax(curDepth + 1, nodeIndex * 2, False, scores, targetDepth),
            minimax(curDepth + 1, nodeIndex * 2 + 1, False, scores, targetDepth)
        )
    else:
        return min(
            minimax(curDepth + 1, nodeIndex * 2, True, scores, targetDepth),
            minimax(curDepth + 1, nodeIndex * 2 + 1, True, scores, targetDepth)
        )
scores = [3, 5, 2, 9, 0, 7]
treeDepth = math.ceil(math.log2(len(scores)))
scores += [0] * (2**treeDepth - len(scores))
print("Optimal value:", minimax(0, 0, True, scores, treeDepth))
print("Leaf evaluations:", count)
```

```python
#Question 2part2
# Leaf evaluation counter
import math
count = 0
def alphabeta(depth, index, maxTurn, scores, target, alpha, beta):
    global count
    if depth == target:
        count += 1
        return scores[index]
    if maxTurn:
        value = float('-inf')
        for i in range(2):
            value = max(
                value,
                alphabeta(depth + 1, index * 2 + i,
                          False, scores, target, alpha, beta)
            )
            alpha = max(alpha, value)
            if alpha >= beta:
                break
        return value
    else:
        value = float('inf')
        for i in range(2):
            value = min(
                value,
                alphabeta(depth + 1, index * 2 + i,
                          True, scores, target, alpha, beta)
            )
            beta = min(beta, value)
            if alpha >= beta:
                break
        return value
scores = [3, 5, 2, 9, 3, 5, 2, 9]
depth = int(math.log2(len(scores)))
result = alphabeta(
    0, 0, True, scores, depth,
    float('-inf'), float('inf')
)
print("Optimal value:", result)
print("Leaf evaluations:", count)
```

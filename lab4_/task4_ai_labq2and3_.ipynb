```python
#Question2
graph = {
    'S': [('A', 1), ('B', 4)],
    'A': [('C', 2)],
    'B': [('G', 5)],
    'C': [('G', 3)],
    'G': []
}

heuristic = {
    'S': 5,
    'A': 4,
    'B': 4,
    'C': 2,
    'G': 0
}

def a_star(start, goal):
    frontier = [(heuristic[start], 0, start, [start])]
    expanded = []
    best_cost = {start: 0}
    while frontier:
        frontier.sort()
        f, g, node, path = frontier.pop(0)
        # Skip old entries
        if g > best_cost[node]:
            continue
        expanded.append(node)
        print("Expanded:", node, "g =", g,
              "h =", heuristic[node], "f =", f)
        if node == goal:
            return path, g, expanded
        for neighbor, cost in graph[node]:
            new_g = g + cost
            if neighbor not in best_cost or new_g < best_cost[neighbor]:
                best_cost[neighbor] = new_g
                new_f = new_g + heuristic[neighbor]
                new_path = path + [neighbor]
                frontier.append(
                    (new_f, new_g, neighbor, new_path)
                )
    return None, float('inf'), expanded

def ucs(start, goal):
    frontier = [(0, 0, start, [start])]
    expanded = []
    best_cost = {start: 0}
    while frontier:
        frontier.sort()
        g, _, node, path = frontier.pop(0)
        
        if g > best_cost[node]:
            continue
        expanded.append(node)
        print("Expanded:", node, "g =", g)
        if node == goal:
            return path, g, expanded
        for neighbor, cost in graph[node]:
            new_g = g + cost
            if neighbor not in best_cost or new_g < best_cost[neighbor]:
                best_cost[neighbor] = new_g
                new_path = path + [neighbor]
                frontier.append(
                    (new_g, new_g, neighbor, new_path)
                )
    return None, float('inf'), expanded
print("----- A* SEARCH -----")
path, cost, expanded = a_star('S', 'G')
print("Returned Path:", path)
print("Path Cost:", cost)
print("Expanded Nodes:", expanded)
print("Expansion Count:", len(expanded))

print("\n----- UCS SEARCH -----")
path, cost, expanded = ucs('S', 'G')
print("Returned Path:", path)
print("Path Cost:", cost)
print("Expanded Nodes:", expanded)
print("Expansion Count:", len(expanded))
```

```python
graph = {
    "a": ["b", "c"],
    "b": ["d", "e"],
    "c": ["f"],
    "d": [],
    "e": [],
    "f": ["g"],
    "g": []
}
def dls(node, goal, limit, depth, path):
    if node in path:
        return "FAILURE", None
    path = path + [node]
    if node == goal:
        return "FOUND", path
    if depth == limit:
        if graph[node]:
            return "CUTOFF", None
        else:
            return "FAILURE", None
    cutoff = False
    for child in graph[node]:
        status, result = dls(
            child, goal, limit, depth + 1, path
        )
        if status == "FOUND":
            return "FOUND", result
        if status == "CUTOFF":
            cutoff = True
    if cutoff:
        return "CUTOFF", None
    return "FAILURE", None
for limit in [1, 2, 3, 4]:
    status, path = dls("a", "g", limit, 0, [])
    print("Limit:", limit)
    print("Status:", status)
    print("Path:", path)
    print("")
```

```python

graph = {
    "a": ["b", "c"],
    "b": ["d", "e"],
    "c": ["f"],
    "d": [],
    "e": [],
    "f": ["g"],
    "g": []
}

visited = []
queue = []

def bfs(node):
    queue.append(node)

    while queue:
        print(queue, visited)

        temp = queue.pop(0)
        visited.append(temp)

        for child in graph[temp]:
            queue.append(child)

bfs("a")

print(queue, visited)

```

```python

graph = {
    "a": ["b", "c"],
    "b": ["d", "e"],
    "c": ["f"],
    "d": [],
    "e": [],
    "f": ["g"],
    "g": []
}

visited = []
stack = []

def dfs(node):
    stack.append(node)

    while stack:
        print(stack, visited)

        temp = stack.pop()

        if temp not in visited:
            visited.append(temp)

            for child in reversed(graph[temp]):
                stack.append(child)

dfs("a")

print(stack, visited)
```
